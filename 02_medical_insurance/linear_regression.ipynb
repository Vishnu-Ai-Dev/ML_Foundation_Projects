```python
import pandas as pd
df = pd.read_csv('insurance.csv')
import numpy as np

df_encoded = pd.get_dummies(df, drop_first=True, dtype=int)

print(df_encoded.head())
print(df_encoded.columns)

X = df_encoded.drop(
    ["charges", "sex_male", "region_northwest", "region_southeast", "region_southwest"],
    axis=1
).values
y=df_encoded['charges'].values


```

```python
#normalization

def normalize(X):
    mu=np.mean(X,axis=0)
    sigma=np.std(X,axis=0)
    x_norm=(X-mu)/sigma
    return x_norm,mu,sigma



```

```python
def prediction(X,w,b):
    pred=np.dot(X,w)+b
    return pred
```

```python
def compute_cost(X,y,w,b):
    m=X.shape[0]
    pred=prediction(X,w,b)
    cost=(1/(2*m))*np.sum(np.square(pred-y))
    return cost
```

```python
def compute_gradient(X,y,w,b):
    m=X.shape[0]
    pred=prediction(X,w,b)
    error=pred-y
    dj_dw=(1/m)*np.dot(X.T,error)
    dj_db=(1/m)*np.sum(error)
    return dj_dw,dj_db
```

```python
def gradient_descent(X, y, w, b, alpha, num_iters):

    for i in range(num_iters):
        dj_dw, dj_db = compute_gradient(X, y,w,b)
        w = w - alpha * dj_dw
        b = b - alpha * dj_db
        if i % 100 == 0:
            cost = compute_cost(X, y, w, b)
            print("Iteration:", i, "Cost:", cost)
    return w, b
```

```python
X_norm, mu, sigma = normalize(X)

print(type(X_norm))
print(X_norm.shape)
print(X_norm[:5])
```

```python
w = np.zeros(X_norm.shape[1])
b = 0.0

print(w)
print(b)
```
