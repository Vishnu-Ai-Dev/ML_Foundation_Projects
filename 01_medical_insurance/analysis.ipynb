```python
import pandas as pd

df = pd.read_csv('insurance.csv')

print("size of the dataset:", df.shape)
print("missing values in the dataset:\n", df.isnull().sum())
print("the first five rows of the file:\n", df.head())

# The Translator for the categorical features
df_encoded = pd.get_dummies(df, dtype=int)

# The Radar of the correlation
correlation_radar = df_encoded.corr()["charges"].sort_values(ascending=False)

print("\n--------FEATURE IMPORTANCE--------")
print(correlation_radar)
```
